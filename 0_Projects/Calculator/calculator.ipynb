## **Description**
I developed a simple calculator application as my first project to practice programming fundamentals and Object-Oriented Programming concepts. The calculator performs basic mathematical operations such as addition, subtraction, multiplication, division, power, and natural logarithm.

```python
import numpy as np

class Calculator:

    def __init__(self):
        pass

    def add(self, val1, val2):
        return val1 + val2

    def sub(self, val1, val2):
        return val1 - val2

    def mul(self, val1, val2):
        return val1 * val2

    def power(self, val1, val2):
        return val1 ** val2

    def log(self, val1):
        return np.log10(val1)

```

```python
# s as a calculator onject
s = Calculator()
```

```python
# Addition
print(s.add(10, 5))
```

```python
# Substraction
print(s.sub(10, 5))
```

```python
# Multiplication
print(s.mul(10, 5))
```

```python
# Exponential
print(s.power(2, 3))
```

```python
# Natural Log
print(s.log(2))
```

#### **1. User will input (3ages).Find the oldest one**

```python
age_1 = int(input("Enter your age"))
age_2 = int(input("Enter your age"))
age_3 = int(input("Enter your age"))
 
def max_age(age_1,age_2,age_3) -> int:
    if age_1 > age_2:
        print(age_1)
    elif age_2 > age_3:
        print(age_2)
    elif age_3 > age_1:
        print(age_3)
    else:
        return "All ages are equal"

oldest_age = max_age(age_1,age_2,age_3)
```

#### ***max Fuction***

```python
def maximum(num1:int,num2:int,num3:int):
    if num1 > num2:
        print(num1)
    elif num2 > num3:
        print(num2)
    elif num3 > num1:
        print(num3)
    else:
        return("All inputs are equal")
```

```python
maximum(22,55,77)
```

#### **2. User will input (2numbers).Write a program to swap the numbers**

#### **4. Write a program that will reverse a four digit number.Also it checks whether the reverse is true.**

```python
num = 1234
reverse = 0


while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10


print(reverse)
```

#### **5. Write a program that will tell whether the number entered by the user is odd or even.**

```python
num = int(input("Enter a number"))

if num % 2 == 0:
    print('Number is Even:',num)
else:
    print("Number is Odd:",num)
```

#### **6. Write a program that will tell whether the given year is a leap year or not.**

```python
year = int(input("Enter a Year: "))

if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("This is a leap year")
else:
    print("This is not a leap year")
```

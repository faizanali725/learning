#### 🟢 **6. — Check Perfect Number**

**A number is called Perfect if the sum of its proper divisors equals the number.**

`For Example:`

- `6 → divisors: 1, 2, 3`
- `1 + 2 + 3 = 6`

- `So: --> 6 → Perfect Number`

***--> Write a program to check whether a user-entered number is perfect or not.***

`Concept: loop + divisor + accumulator.`

```python
num = int(input("Enter a Number: "))

deviser_sum = 0

for i in range(1,num):
    if num % i == 0:
        deviser_sum +=1

if deviser_sum == num and num > 0:
    print(f'This is Perfect Number: {num}')
else:
    print(f'This is Not Perfect Number: {num}')
```

#### 🔴 **7. — Second Largest Without Sorting**

***---> Given:***

`numbers = [10, 5, 20, 8, 15, 20]`

***---> Rules:***

- `❌ Don't use sort()`
- `❌ Don't use sorted()`
- `❌ Don't use set()`

`Concept: largest + second_largest + comparison.`

```python
numbers = [10, 5, 20, 8, 15, 20]

largest = numbers[0]
second_largest = numbers[0]

for i in numbers:
    if i > largest:
        second_largest = largest
        largest = i

    elif i > second_largest and i != largest:
        second_largest = i

print("Second Largest:", second_largest)
```

#### 🔴 **5. — Challenge 🔥**

***--> Function:***

- `analyze_number(number)`

`Number ke:`

- `digit count`
- `digit sum`
- `even digits count`
- `odd digits count`

`Yahan tumhe ek function mein multiple logic patterns combine karne honge.`

```python
num = int(input("Enter Number: "))

def analyze_number(number):

    digit_count = 0
    digit_sum = 0
    even_count = 0
    odd_count = 0

    while number > 0:
        last_digit = number % 10
        digit_sum = digit_sum + last_digit
        number = number // 10

        digit_count += 1

        if last_digit % 2 == 0:
            even_count += 1
        else:
            odd_count += 1

    return ('Digit Sum: ',digit_sum,'Even Number Count: ',even_count,'Odd Number Count: ',odd_count,'Total Digit: ',digit_count)

result = analyze_number(num)
print(result)
        
```

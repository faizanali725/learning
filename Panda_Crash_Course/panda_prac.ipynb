## **----> Pandas Crash Course <----**

```python
import pandas as pd
```

## **1. Loading Dataset**

```python
df = pd.read_csv("retail_sales.csv")

print("Dataset Loaded")

display(df.head())
```

## **2. Dataset Overview**

```python
print("\nDataset Shape:\n")

print(df.shape)

print("\nColumn Names:\n")

print(df.columns)

print("\nData Types:\n")

print(df.dtypes)

print("\nSummary Statistics:\n")

print(df.describe())
```

## **3. Selecting Columns**

```python
print("Selecting Single Column:\n")

df[["Rating"]].tail()
```

```python
print("Selecting Multiple Column:\n")

df[["Order_ID", "Quantity", "Price", "Rating"]].tail()
```

## **4. Filtering Data**

```python
print("Orders where Category = Electronics")

electronic = df[df["Category"] == "Electronics"]

electronic[["Order_ID", "Date", "Customer", "City"]].head()
```

***Multiple Conditions***

```python
high_value_orders = df[(df["Price"]>10000) & (df["Category"] == "Electronics")]
high_value_orders.head(2)
```

## **5. Adding New Columns**

```python
df["Total_Sales"] = df["Quantity"] * df["Price"]

print("Total Sales Column Added: ")

df.tail(2)
```

## **6. Sorting Data**

```python
sorted_df = df.sort_values(by="Price",ascending=False).head()
sorted_df
```

## **7. Handling Missing Values**

```python
df.head().isnull()
```

```python
df.isnull().sum()
```

## **8. Unique Values**

```python
df["Category"].unique()
```

```python
df["City"].nunique()
```

## **9. Value Counts**

```python
df["Category"].value_counts()
```

## **10. GroupBy Operations**

```python
df.groupby("Category")["Total_Sales"].sum()
```

***Multiple Aggregations***

```python
category_stats = df.groupby("Category")["Total_Sales"].agg(["sum","mean","max"])
category_stats
```

## **11. Group By City**

```python
city_stats = df.groupby("City")["Total_Sales"].sum()
city_stats
```

## **12. Pivot Table**

```python
pivot = pd.pivot_table(
    df,
    values="Total_Sales",
    index="City",
    columns="Category",
    aggfunc="sum"
)

print("\nPivot Table:")
pivot
```

## **13. Date Operations**

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Month"] = df["Date"].dt.month
df.head()

monthly_sales = df.groupby("Month")["Total_Sales"].sum()

print("\nMonthly Sales")

monthly_sales
```

## **14. Top Customers**

```python
top_customers = df.groupby("Customer")["Total_Sales"].sum().sort_values(ascending=False)

print("\nTop Customers:")

top_customers.head()
```

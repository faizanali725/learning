
# ***🚀 Streamlit – Complete Notes***

## **1. What is Streamlit?**

Streamlit is an open-source Python framework used to create interactive web applications, especially for Data Science, Machine Learning, and AI projects.

- Build web apps using Python.
- No HTML, CSS, or JavaScript knowledge required.
- Create interactive dashboards.
- Display data, charts, and visualizations.
- Share your applications easily.

## **8. Basic Streamlit Application**

```python
import streamlit as st

st.title("My First Streamlit App")

st.header("Welcome!")

name = st.text_input("Enter your name")

if st.button("Submit"):
    st.success(f"Welcome {name}!")
```

## **10. Important Things to Remember**

- Streamlit applications are written in Python.
- Every time a widget changes, Streamlit reruns the script.
- Use `st.session_state` to preserve data between reruns.
- Use `st.sidebar` for navigation and filters.
- Use Pandas for data manipulation.
- Use Matplotlib and Plotly for visualizations.

## **13. Session State**

***st.session_state*** is used to store information between Streamlit reruns.

`--> Example:`

```python
import streamlit as st

if "count" not in st.session_state:
    st.session_state.count = 0

if st.button("Click Me"):
    st.session_state.count += 1

st.write("Count:", st.session_state.count)
```

## **14. Caching**

***Caching*** improves application performance by avoiding unnecessary repeated work.

| **Command** | **Purpose** |
|---|---|
| `@st.cache_data` | Cache data and calculations |
| `@st.cache_resource` | Cache resources like models and connections |

## **15. Forms**

***Forms*** allow users to enter multiple inputs and submit them together.

```python
with st.form("my_form"):
    name = st.text_input("Name")
    age = st.number_input("Age")

    submit = st.form_submit_button("Submit")

if submit:
    st.write(name, age)
```

```python
class Atm:

    def __init__ (self):

        self.pin = ""
        self.balance = 0

        self.menu()

    def get_pin(self):
        return self.__pin
        
    def set_pin(self,New_pin):
        if type(New_pin) == str:
            self.__pin = New_pin
            print("Pin Changed")
        else:
            print("Not Allowed")

    def menu(self):
        user_input = input("""
                        Hello,How would you like to proceed?
                        1. Enter 1 to Create Pin
                        2. Enter 2 to Deposit
                        3. Enter 3 to Withdraw
                        4. Enter 4 to Check Balance
                        5. Enter 5 to Exit
""")
        if user_input == "1":
            self.create_pin()
        elif user_input =="2":
            self.deposit()
        elif user_input == "3":
            self.withdraw()
        elif user_input == "4":
            self.check_balance()
        else:
            print("Bye")



    def create_pin(self):
        self.pin = input("Enter your Pin")
        print("Pin set Successfully")


    def deposit(self):
        pin = input("Enter your Pin")
        if pin == self.pin:
            amount = int(input("Enter your amount"))
            self.balance = self.balance + amount
            print("Deposit Successful")
        else:
            print("Invalid Pin")

    
    def withdraw(self): 
        pin = input("Enter your Pin")
        if pin == self.pin:
            amount = int(input("Enter your amount"))
            if amount < self.balance:
                self.balance = self.balance - amount
                print("Withdraw Successful")
            else:
                print("Insufficient Funds")
        else:
            print("Invalid Pin")  


    def check_balance(self):
        pin = input("Enter your Pin")
        if pin == self.pin:
            print(self.balance)
        else:
            print("Invalid Pin")

            
```

```python
class Fraction:

    def __init__(self,n,d):
        self.num = n
        self.den = d



    def __str__(self):
        return "{}/{}".format(self.num,self.den)

    

    def __add__(self,other):

        temp_num = self.num * other.den + other.num * self.den
        temp_den = self.den * other.den

        return "{}/{}".format(temp_num,temp_den)

    

    def __sub__(self,other):
        
            temp_num = self.num * other.den - other.num * self.den
            temp_den = self.den * other.den
    
            return "{}/{}".format(temp_num,temp_den)

    

    def __mul__(self,other):
        
            temp_num = self.num * other.num
            temp_den = self.den * other.den
    
            return "{}/{}".format(temp_num,temp_den)


    
    def __truediv__(self,other):
            
                temp_num = self.num * other.den
                temp_den = self.den * other.num
        
                return "{}/{}".format(temp_num,temp_den)
```

```python
class Student:
    
    def __init__(self):
        self.name = ""
        self.age = 0
        self.marks = 0

    def student_name(self):
        self.name = input("Enter your Name: ")
        print("Name set successfully!")

    def student_age(self):
        self.age = int(input("Enter your Age: "))
        print("Age set successfully!")

    def student_marks(self):
        self.marks = int(input("Enter your Marks: "))
        print("Marks set successfully!")

    def student_result(self):
        if self.marks >= 40:
            print("Result: Pass")
        else:
            print("Result: Fail")

    def student_grades(self):
        if self.marks >= 90:
            print("Grade A+ : " , self.marks)
        elif self.marks >= 80:
            print("Grade A : " , self.marks)
        elif self.marks >= 70:
            print("Grade B : " , self.marks)
        elif self.marks >= 60:
            print("Grade C : " , self.marks)
        elif self.marks >= 50:
            print("Grade D : " , self.marks)
        elif self.marks >= 40:
            print("Grade E : " , self.marks)
        else:
            print("Fail")

    def display_info(self):
        print("\n----- Student Information -----")

        print("Name:", self.name)
        print("Age:", self.age)
        print("Marks:", self.marks)


# Object creation
student1 = Student()

# Taking student information
student1.student_name()
student1.student_age()
student1.student_marks()

# Display information
student1.display_info()

# Check result
student1.student_result()
student1.student_grades()

print("--------->>>>------------")    


student2 = Student()

# Taking student information
student2.student_name()
student2.student_age()
student2.student_marks()

# Display information
student2.display_info()

# Check result
student2.student_result()
student2.student_grades()
```

```python
class Student:
    
    def __init__(self):
        self.name = ""
        self.age = 0
        self.marks = 0

    def student_name(self):
        self.name = input("Enter your Name: ")

    def student_age(self):
        self.age = int(input("Enter your Age: "))

    def student_marks(self):
        self.marks = int(input("Enter your Marks: "))

    def student_result(self):
        if self.marks >= 40:
            return "Pass"
        else:
            return "Fail"

    def student_grade(self):
        if self.marks >= 90:
            return "A+"
        elif self.marks >= 80:
            return "A"
        elif self.marks >= 70:
            return "B"
        elif self.marks >= 60:
            return "C"
        elif self.marks >= 50:
            return "D"
        elif self.marks >= 40:
            return "E"
        else:
            return "Fail"

    def display_info(self):
        print("\n----- Student Information -----")
        print("Name:", self.name)
        print("Age:", self.age)
        print("Marks:", self.marks)
        print("Result:", self.student_result())
        print("Grade:", self.student_grade())


# Empty list for storing students
students = []

# Ask how many students
number = int(input("How many students do you want to add? "))

# Create students
for i in range(number):

    print(f"\n----- Student {i + 1} -----")

    student = Student()

    student.student_name()
    student.student_age()
    student.student_marks()

    students.append(student)


# Display all students
print("\n========== ALL STUDENTS ==========")

for student in students:
    student.display_info()
```

```python
# This is Fraction addition not mathematical addition is called Operator Overloading
class Fraction:

    def __init__(self,n,d):
          
        self.num = n
        self.den = d

    def __add__(self,other):
        
            temp_num = self.num * other.den + other.num * self.den
            temp_den = self.den * other.den

            return "{}/{}".format(temp_num,temp_den)

a = Fraction(4,5)
b = Fraction(6,7)
print(a + b)
```

***---> super() keyword***

Example 1 --> `User super() keyword`

`--> super() keyword not work out of class`

```python
class Phone:
    
    def __init__(self,price,brand,camera):

        print("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera


    def buy(self):
        print("Buying a phone")

class SmartPhone(Phone):

    def buy(self):
        print("Buying a smartphone")
        super().buy()  # -----> This is super() keyword this use for access parent method & constructor

s = SmartPhone(20000,"Infinix",15)

s.buy()
```

Example 2 --> `super() keyword use with parent constructor`

```python
class Phone:
    
    def __init__(self,price,brand,camera):

        print("Inside phone constructor")
        self.price = price
        self.brand = brand
        self.camera = camera


class SmartPhone(Phone):

    def __init__(self, price, brand, camera,os,ram):
        print("First of all")
        super().__init__(price, brand, camera)
        self.os = os
        self.ram = ram

        print("Inside smartphone constructor")

s = SmartPhone(100000,"Apple",15,"Android","8gb")

print(s.os)
print(s.ram)
print(s.price)
print(s.brand)
```

Example 3 --> `On super() keyword`

`--> super() keyword in constructor should be first statement`

```python
class Parent:

    def __init__(self,num):
        self.__num = num


    def get_num(self):
        return self.__num

class Child(Parent):

    def __init__(self, num,val):
        super().__init__(num)
        self.__val = val


    def get_val(self):
        return self.__val

s = Child(100,200)
print(s.get_num())
print(s.get_val())
        
```

Example 4 --> `super() keyword`

```python
class Parent:

    def __init__(self):
        self.__num = 150

    def show(self):
        print("Parent:",self.__num)

class Child(Parent):

    def __init__(self):
        super().__init__()
        self.__val = 100

    def show(self):
        print("Child:",self.__val)

dad = Parent()
dad.show()

son = Child()
son.show()
```

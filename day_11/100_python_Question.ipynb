#### **1. User will input (3ages).Find the oldest one**

```python
age_1 = int(input("Enter your age"))
age_2 = int(input("Enter your age"))
age_3 = int(input("Enter your age"))
 
def max_age(age_1,age_2,age_3) -> int:
    if age_1 > age_2:
        print(age_1)
    elif age_2 > age_3:
        print(age_2)
    elif age_3 > age_1:
        print(age_3)
    else:
        return "All ages are equal"

oldest_age = max_age(age_1,age_2,age_3)
```

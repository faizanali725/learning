## **1. --> What is Iteration**

- Iteration is a general term for taking item of something, one after another. Any time you use a loop, ecplicit or implicit, to go over a group of items, that is iteration.

```python
num = [1,2,3,4]

for i in num:
    print(i)
```

## **2. --> What is Iterator**

- An iterator is an object that allows the programmer to traverse through a sequence of data without having to store the entire data in the memory.

```python
L = [x for x in range(1,100000)]

# for i in L:
#     print(i*2)

import sys

print(sys.getsizeof(L)/1024)

# iterator
x = range(1,10000000000)

# for i in x:
#     print(i*2)

print(sys.getsizeof(x)/1024)  
```

## **3. --> What is Iterable**

- Iterable is an object, which one can iterate over 
- It generates an iterator when passed to iter() method.

***--> Point to Remember***

- Every **Iterator** is also and **Iterable**
- Not all **Iterable** are **Iterators**

***--> Trick***

- Every **Iterable** has an **iter() Function**
- Every **Iterator** has both **iter() Function** as well as a **Next Function**

```python
# Take integer value and check can this is Iterable
a = 2

dir(a)

# Find iter() method Not find, so this is not Iterable
```

- Tuple, Sets, and Dictonary is Iterable

```python
# Take a Tuple and check can this is Iterable
T = (1,2,3)

dir(T)

# Yes Find Iter() method this is Iterable
```

--> How to check which object is Iterator 
- Use dir() Function and see can has **iter and Next function**.

```python
L = {1,2,3,4}

# L iter has but Not Next function, So this is Not an Iterator
dir(L)


# It generates an iterator when passed to iter() method.
iter_L = iter(L)
# Now iter_L is an Iterator
dir(iter(L))
```

## **5. --> Making our own for loop**

```python
def My_for_loop(iterable):

    iterator = iter(iterable)

    while True:

        try:
            print(next(iterator))
        except StopIteration :
            break
```

```python
a = [1,2,3]
My_for_loop(a)
```

```python
b = (1,2,3)
My_for_loop(b)
```

```python
c = range(1,11)
My_for_loop(c)
```

## **--> Lets create our own range() Function**

```python
class My_range:

    def __init__(self,start,end):
        self.start = start
        self.end = end

    def __iter__(self):
        return My_range_iterator(self)


class My_range_iterator:

    def __init__(self,iterable_obj):
        self.iterable = iterable_obj

    def __iter__(self):
        return self

    def __next__(self):
        if self.iterable.start >= self.iterable.end:
            raise StopIteration

        current = self.iterable.start
        self.iterable.start +=1
        return current
```

```python
for i in My_range(1,11):
    print(i)
```

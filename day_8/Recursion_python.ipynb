## **1. --> Recursion**

`Iterative Vs Recursive`

`a*b`

***--> Iterative***

```python
def multiply(a,b):
    result = 0

    for i in range(b):
        result = result + a

    print(result)

multiply(3,4)
```

```python
multiply(5,5)
```

***--> Recursive***

```python
def mul(a,b):
    if b == 1:
        return a
    else:
        return a + mul(a,b - 1)
print(mul(5,6))
```

```python
mul(4,5)
```

## **2. --> Factorial**

```python
def fact(num):
    if num == 1:
        return 1
    else:
        # 5 = 5 * 4!
        return num * fact(num - 1)

print(fact(5))
```

```python
fact(6)
```

## **3. --> Palindrome**

```python
def palin(text):
    if len(text) == 1:
        print("palindrome")
    else:
        if text[0] == text[-1]:
            palin(text[1:-1])
        else:
            print("Not a Palin")

palin("madam")
palin("lahore")
palin("faizan")
palin("malayalam")
```

## **4. --> The Rabbit Problem**

`This method of fibonacci`

***--> Fibonacci***

```python
def fib(m):
    if m == 0 or m == 1:
        return 1
    else:
        return fib(m-1) + fib(m-2)

print(fib(12))
```

```python
fib(24)
```

```python
fib(10)
```

```python
fib(36)
```

## **5. --> Memoization**

```python
import time
def memo(m,d):
    if m in d:
        return d[m]
    else:
        d[m] =  memo(m-1,d) + memo(m-2,d)
        return d[m]
   
start = time.time()
d = {0:1, 1:1}
print(memo(500,d))
print(time.time()-start)
```

## **2. --> HOF - 'Higher Order Function'**

```python
def return_sum(func,L):

    result = 0

    for i in L:

        if func(i):
            result = result + i

    return result


L = [11,22,34,66,87,67,56,89,67,32,35]

x = lambda x: x%2 == 0
y = lambda x: x%2!= 0
z = lambda x: x%3 == 0


print(return_sum(x,L))
print(return_sum(y,L))
print(return_sum(z,L))
```

## **3. --> Map**

```python
L = [1,2,3,4,5,6]

map(lambda x:x*2,L)
list(map(lambda x:x*2,L))
```

```python
list(map(lambda x: x%2 == 0,L))
```

```python
student = [{'name':'faizan','age':17},{'name':'robert berlt','gender':'M'},{'name':'goseph'}]
list(map(lambda student:student['name'],student))
```

```python
student = [
    {
        'name':'faizan','age':17
        }
    ,{
        'name':'robert berlt','gender':'M'
        }
    ,{
        'name':'goseph'
        }
    ]
list(map(lambda student:student['name'],student))
```
